### Mann Whitney U test
Непараметрический тест, который используется для сравнения двух независимых выборок. Он помогает определить, есть ли статистически значимые различия между двумя группами. Этот тест часто применяется в случаях, когда данные не соответствуют нормальному распределению, когда размеры выборок малы, или при ординальных данных.

Нулевая гипотеза (H0): Распределения двух групп идентичны (или, эквивалентно, медианы двух групп равны).<br>
Альтернативная гипотеза (H1): Распределения двух групп различны (или медианы двух групп различны).

Сначала ранжируются все данные, объединенные из обеих групп (Ранжирование может быть выполнено с учетом повторяющихся значений).<br>
Затем вычисляется сумма рангов для каждой группы.<br>
Используя сумму рангов, рассчитывается статистика U, которая указывает на степень различия между группами.

#### Пример задачи
У нас есть оценки удовлетворенности клиентов из двух разных регионов, и мы хотим определить, существует ли значительная разница в уровнях удовлетворенности между этими двумя регионами.

```python
import numpy as np
from scipy.stats import mannwhitneyu
```

```python
# Назначаем random seed
np.random.seed(105)

# Синтезируем данный для Региона А
region_a = np.append(np.random.randint(1, 4, 25), np.random.randint(4, 6, 25))

# Синтезируем данный для Региона B
region_b = np.append(np.random.randint(1, 4, 18), np.random.randint(4, 6, 32))
```

```python
# Проводим Mann Whitney test
u_statistic, p_value = mannwhitneyu(region_a, region_b, alternative='less')
```

```python
# Смотрим результаты
print("\nMann-Whitney U тест результаты:")
print("U Statistic:", u_statistic)
print("P-Value:", p_value)
```

```python
# Интерпретация
alpha = 0.1
if p_value < alpha:
    print("\nВывод: существует значительная разница в показателях удовлетворенности между регионом А и регионом Б.")
else:
    print("\nВывод: существенной разницы в показателях удовлетворенности между регионом А и регионом Б нет.")
```

### Kolmogorov-Smirnov test
Статистический тест, который используется для проверки того, насколько хорошо распределение данных соответствует теоретически предполагаемому распределению. Тест помогает определить, насколько эмпирическое распределение данных отличается от теоретического распределения (например, нормального распределения) или проверить, совпадают ли два эмпирических распределения.

**One-sample test**<br>
<img src="data/KS1.png" alt="One-sample test" width="500"/>

```python
from scipy.stats import kstest
import numpy as np
```

```python
np.random.seed(105)
sample = np.random.normal(loc=0, scale=1, size=100)
```

```python
ks_statistic, p_value = kstest(sample, 'norm')
```

```python
print(f"KS Statistic: {ks_statistic}")
print(f"P-value: {p_value}")
```

**Two-sample test**<br>
<img src="data/KS2.png" alt="One-sample test" width="500"/>

```python
from scipy.stats import ks_2samp
import numpy as np
```

```python
np.random.seed(105)
sample1 = np.random.normal(loc=0, scale=1, size=400)
sample2 = np.random.normal(loc=0, scale=1, size=500)
```

```python
ks_statistic, p_value = ks_2samp(sample1, sample2)
```

```python
print(f"KS Statistic: {ks_statistic}")
print(f"P-value: {p_value}")
```

### Shapiro-Wilk test
статистический тест, который используется для проверки гипотезы о нормальности распределения данных. Другими словами, он помогает определить, насколько хорошо данные следуют нормальному распределению.

<img src="data/shapiro.png" alt="One-sample test" width="500"/>

```python
import numpy as np
from scipy.stats import shapiro
```

```python
data = np.array([2.3, 2.5, 2.7, 3.0, 3.1, 3.4, 3.6, 3.8, 4.0, 4.1])
```

```python
statistic, p_value = shapiro(data)
```

```python
print("Shapiro-Wilk Test Statistic:", statistic)
print("P-Value:", p_value)
```

```python
alpha = 0.05
if p_value > alpha:
    print("Не удалось отклонить null hypohtesis - дата распределена нормально")
else:
    print("Отклоняем null hypothesis - дата не распределена нормально")
```

### Wilcoxon test

непараметрический статистический тест, используемый для сравнения связанных (зависимых) выборок. Он проверяет, отличается ли медианная разница между парными наблюдениями от нуля. Работает с ординальными значениями

Нулевая гипотеза (H0): Медианная разница между парными наблюдениями равна нулю.<br>
Альтернативная гипотеза (H1): Медианная разница не равна нулю (двусторонний тест) или больше/меньше нуля (односторонний тест).

Алгоритм просчета:
* для кажой пары считаем абсолютную разницу
* ранжируем получившиеся разницы, присваевам ранг
* считаем сумму положительных и отрицательных рангов
* считаем W статистику и стандартное отколнение
* получивашиеся статитсики можно применить для нормального распределения (при достаточно большой выборке n>25)
* при маленькой выборке используются табличные (имперические) значения
* стандартное отклонение корректируется при одинковых значениях в двух выборках

```python
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from scipy.stats import norm
```

```python
np.random.seed(116)
sample_pd = pd.DataFrame({'val1': np.random.randint(20, 30, 30),
                          'val2': np.random.randint(21, 31, 30)}
                        )
sample_pd.head()
```

```python
#вычисление разности
sample_pd['diff'] = sample_pd.apply(lambda x: x['val2'] - x['val1'], axis=1)
sample_pd.head()
```

```python
#вычисление абсолютной разницы
sample_pd['abs_diff'] = abs(sample_pd['diff'])
sample_pd.head()
```

```python
#арнжирование абсолютных значений
sample_pd['rank'] = sample_pd['abs_diff'].rank(method='average', ascending=False)
sample_pd.head()
```

```python
#сумма положительных рангов
w_pos = sample_pd[sample_pd['diff']>0]['rank'].sum()
w_pos
```

```python
#сумма негативных рангов
w_neg = sample_pd[sample_pd['diff']<0]['rank'].sum()
w_neg
```

```python
#w статистика - это наименьшее из двух статистик
w = min(w_pos, w_neg)
w
```

```python
#повторяющиеся значения
(sample_pd
 .groupby('diff', as_index=False)
 .agg(cnt = ('rank', 'count'))
)
```

```python
sample_pd.shape
```

```python
# #стандартное отколнение
n = 30
t2 = 2
t4 = 2
t6 = 2
t7 = 5
t8 = 4
t9 = 2
t10 = 3
t12 = 2
std = np.sqrt(n * (n + 1) * (2*n + 1) / 24 - 
              t2 * (t2 - 1) * (2*t2 + 1) / 24 -
              t4 * (t4 - 1) * (2*t4 + 1) / 24 -
              t6 * (t6 - 1) * (2*t6 + 1) / 24 -
              t7 * (t7 - 1) * (2*t7 + 1) / 24 -
              t8 * (t8 - 1) * (2*t8 + 1) / 24 -
              t9 * (t9 - 1) * (2*t9 + 1) / 24 -
              t10 * (t10 - 1) * (2*t10 + 1) / 24 -
              t12 * (t12 - 1) * (2*t12 + 1) / 24)
std
```

```python
#W при нулевой гипотезе
W_mean = n * (n+1) / 4
W_mean
```

```python
#считаем Z статистику
z = (w - W_mean) / std
z
```

```python
#считаем вероятность p
p = 2 * (1 - norm.cdf(abs(z)))
p
```

```python
stat, p = wilcoxon(sample_pd['val1'], sample_pd['val2'])
stat, p
```
