```python
#1. Создайте функцию f(x) = 3 * x^2 + 2 * x - 10
```

```python
def f(x):
    return 3 * x ** 2 + 2 * x - 10

print(f(1)) #Для проверки правильности подсчета
```

```python
#2. Создайте список x_points (или массив через numpy), в котором у нас лежат числа от -10 до 10
```

```python
x_points = []

for i in range (-10, 11):
    x_points.append(i)

print(x_points)
```

```python
#3. Используйте x_points как аргумент для созданной функции f(x), и выведите на экран результаты функции для каждого значения 
# x_points, также сохраните результаты в y_points (cписок или массив)
```

```python
y_points = []

for i in x_points:
    point = f(i)
    y_points.append(point)

print(y_points)
```

```python
#4. Покажите на графике x_points & y_points
```

```python
import matplotlib.pyplot as plt

plt.plot(x_points, y_points, marker = 'o', linestyle = '--')
plt.show()
```

```python
#5. Найдите производную функции f(x) и создайте новую функцию на ее основе как f_diff(x)
```

```python
import sympy as sp

x = sp.symbols('x')
f_x = 3 * x ** 2 + 2 * x - 10
f_diff = sp.diff(f_x, x)
print(f_diff)

def f_diff(x):
    return 6*x + 2
```

```python
#6 Чему равна производная при x = 5
```

```python
f_diff(5)
```

```python
#7. Покажите на графике f(x) & f_diff(x)
```

```python
y_diff_points = []

for i in x_points:
    point = f_diff(i)
    y_diff_points.append(point)

plt.plot(x_points, y_points, marker = 'o', linestyle = '--', label = 'f(x)')
plt.plot(x_points, y_diff_points, marker = 'o', linestyle = '--', label = 'f_diff(x)')
plt.legend()
plt.show()
```

```python
#10. Найдите мин или макс функции f(x) и покажите на графике касательную линию, которая проходит через этот мин или макс функции
```

```python
#Уравнение производной: 6*x + 2
#6x + 2 = 0
#6x = -2
#x = -2/6
#x = -1/3

x0 = -1/3
y0 = f(x0)

y0_tangent = []

for i in x_points:
    y0_tangent.append(y0)

plt.plot(x_points, y_points, marker = 'o', linestyle = '--', label = 'f(x)')
plt.plot(x_points, y0_tangent, label = 'Минимум')
plt.legend()
plt.show()
```
