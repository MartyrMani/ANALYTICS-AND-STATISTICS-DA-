```python
#1. Создайте функцию f(x) = x^6 + 10 * x^3 - 1000
```

```python
def f(x):
    return x ** 6 + 10 * x ** 3 - 1000

print(f(1)) #Для проверки правильности подсчета
```

```python
#2. Создайте список x_points (или массив через numpy), в котором у нас лежат числа от -10 до 10
```

```python
x_points = []

for i in range (-10,11):
    x_points.append(i)

print(x_points)
```

```python
#3. Используйте x_points как аргумент для созданной функции f(x), и выведите на экран результаты функции для каждого значения 
# x_points, также сохраните результаты в y_points (cписок или массив)
```

```python
y_points = []

for i in x_points:
    point = f(i)
    y_points.append(point)
    print(point)

print()
print(y_points)
```

```python
#4. Покажите на графике x_points & y_points
```

```python
import matplotlib.pyplot as plt

plt.plot(x_points, y_points, marker = 'o', linestyle = '--')
plt.show()
```

```python
#5. Найдите производную функции f(x) и создайте новую функцию на ее основе как f_diff(x)
```

```python
import sympy as sp

x = sp.symbols('x')
f = x ** 6 + 10 * x ** 3 - 1000
f_diff = sp.diff(f, x)
print(f_diff)

def f_diff(x):
    return 6 * x ** 5 + 30 * x ** 2
```

```python
#6 Чему равна производная при x = 10
```

```python
f_diff(10)
```

```python
#7 Покажите на графике f(x) & f_diff(x)
```

```python
y_diff_points = []

for i in x_points:
    point = f_diff(i)
    y_diff_points.append(point)
    print(point)

plt.plot(x_points, y_points, marker = 'o', linestyle = '--', label='f(x)')
plt.plot(x_points, y_diff_points, marker = 'o', linestyle = '--', label='f_diff(x)')
plt.legend()
plt.show()
```

```python
#8 Найдите уравнение касательной линии в точке x = 9
```

```python
#(y-y1)=m*(x-x1)
x1 = 9
y1 = f(9) #537731
m = f_diff(9) #356724
#y - 537731 = 356724 * x - 356724 * 9
#y = 356724 * x - 2672785

x_tangent = []
y_tangent = []

for i in range(8, 11):
    x_tangent.append(i)
    y_tangent.append(356724 * i - 2672785)

plt.plot(x_tangent, y_tangent, marker = 'o', linestyle = '--')
plt.show()
```

```python
#9 Покажите на графике функцию f(x) & касательную линию в точке x = 9
```

```python
plt.plot(x_points, y_points, marker = 'o', linestyle = '--', label='f(x)')
plt.plot(x_tangent, y_tangent, marker = 'o', linestyle = '--', label='Tangent Line')
plt.legend()
plt.show()
```
