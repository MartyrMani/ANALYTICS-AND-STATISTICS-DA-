```python
import pandas as pd
from scipy.stats import mannwhitneyu, kstest, shapiro, ks_2samp
import matplotlib.pyplot as plt
```

#### Задание 1
Вам подготовили датасет в файле compare_df.csv<br>
В нем есть всего две колонки sample1 и sample2<br>
Проведите Mann U Whitney тест чтобы проверить, идентично ли распределение двух этих сэмплов с уровнем значимости в 95%

```python
compare_df = pd.read_csv('compare_df.csv')

print(compare_df.head())

sample1 = compare_df['sample1']
sample2 = compare_df['sample2']

u_statistic, p_value = mannwhitneyu(sample1, sample2, alternative = 'two-sided')
print("\nU Statistic: ", u_statistic)
print("P Value: ", p_value)

alpha = 0.05
if p_value < 0.05:
    print("\nОтвергаем нулевую гипотезу.")
    print("Вывод: существенная разница между двумя колонками существует.")
else:
    print("\nНе отвергаем нулевую гипотезу.")
    print("Вывод: существенная разница между двумя колонками не существует.")
```

#### Задание 2
Используя те же самые два сэмпла из compare_df, проведите Kolmogorov Smirnov тест для проверки идентичности распределений

```python
ks_statistic, p_value = ks_2samp(sample1, sample2)

print("KS Statistic: ", ks_statistic)
print("P Value: ", p_value)

alpha = 0.05
if p_value < 0.05:
    print("\nОтвергаем нулевую гипотезу.")
    print("Вывод: существенная разница между двумя колонками существует.")
else:
    print("\nНе отвергаем нулевую гипотезу.")
    print("Вывод: существенная разница между двумя колонками не существует.")
```

#### Задание 3
Вам подготовили датасет в файле normality_test.csv<br>
В нем всего одна колонка sample<br>
Проведите Kolmogorov Smirnov тест, чтобы проверить нормальность данных с уровнем значимости в 95%

```python
sample = pd.read_csv('normality_test.csv')
sample_data = sample['sample']

ks_statistic, p_value = kstest(sample_data, 'norm')

print("KS Statistic: ", ks_statistic)
print("P Value: ", p_value)

alpha = 0.05
if p_value < 0.05:
    print("\nОтвергаем нулевую гипотезу.")
else:
    print("\nНе отвергаем нулевую гипотезу.")
```

#### Задание 4
Используя те же самые данные, проведите тест на нормальность используя Shapiro Wilk тест<br>
Какому тесту (ШапироУилк или КолмогоровСмирнов) вы доверяете больше и чей результат выберите?

```python
sh_statistic, p_value = shapiro(sample)

print("Shapiro Statistic: ", sh_statistic)
print("P Value: ", p_value)

alpha = 0.05
if p_value  > 0.05:
    print("\nНе отвергаем нулевую гипотезу.")
else:
    print("\nОтвергаем нулевую гипотезу.")

#Шапиро–Уилка, так как он специально предназначен для проверки нормальности распределения 
#и обычно обладает большей чувствительностью для таких задач.
```
