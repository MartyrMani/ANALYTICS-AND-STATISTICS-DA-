```python
from scipy.stats import norm
import numpy as np
import random
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
mean_pages = 20
std_pages = 2

threshold_pages = 18

probability_pages = norm.cdf(threshold_pages, loc = mean_pages, scale = std_pages)

print(f"Вероятность что брошюра будет содержать менее {threshold_pages} страниц составляет: {probability_pages:.4f}")
```

```python
np.random.seed(105)
height_sample = np.random.normal(175, 6, 1000)

ul = norm.ppf(0.975, 175, 6)
ll = norm.ppf(0.025, 175, 6)

sns.kdeplot(height_sample)
plt.axvline(ul, linestyle = '--', color = "red")
plt.axvline(ll, linestyle = '--', color = "red")
plt.show()
```

```python
np.random.seed(105)
grades_sample = np.random.normal(75, 10, 100)

print("Точечная оценка составляет: ", np.mean(grades_sample))
```
