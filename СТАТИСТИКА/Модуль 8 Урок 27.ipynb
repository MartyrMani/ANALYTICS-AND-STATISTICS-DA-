```python
import random
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import statsmodels.api as sm
```

```python
list_of_dice = [random.randint(1, 6) for i in range(1000)]
#print(list_of_dice)

plt.hist(list_of_dice, bins = [1, 2, 3, 4, 5, 6], rwidth = 0.9, density = True)
plt.title("Вероятность цифры кубика")
plt.ylabel("Вероятность")
plt.show()
```

```python
np.random.seed(105)
normal_data = np.random.normal(50, 15, 10000)

mean = np.mean(normal_data)
std = np.std(normal_data)

plt.hist(normal_data, rwidth = 0.9, bins = np.arange(0, 105, 5), density = True)
plt.axvline(mean, linestyle = '-', color = "black", label = "mean")
plt.axvline(mean + std, linestyle = '--', color = "red", label = "+- 1std")
plt.axvline(mean - std, linestyle = '--', color = "red")
plt.axvline(mean + std * 2, linestyle = '--', color = "purple", label = "+- 2std")
plt.axvline(mean - std * 2, linestyle = '--', color = "purple")
plt.axvline(mean + std * 3, linestyle = '--', color = "green", label = "+- 3 std")
plt.axvline(mean - std * 3, linestyle = '--', color = "green")
plt.title("Вероятность нормального распределения")
plt.ylabel("Вероятность")
plt.legend()
plt.show()

outliers = np.random.normal(200, 5, 10)
data_with_outliers = np.concatenate((normal_data, outliers))

new_mean = np.mean(data_with_outliers)
new_std = np.std(data_with_outliers)

print("Исходное среднее: ", mean)
print("Новое среднее: ", new_mean)
print("Исходное стандартное отклонение: ", std)
print("Новое стандартное отклонение: ", new_std)

plt.hist(data_with_outliers, rwidth = 0.9, bins = np.arange(0, 220, 5), density = True)
plt.axvline(new_mean, linestyle = '-', color = "black", label = "mean")
plt.axvline(new_mean + new_std, linestyle = '--', color = "red", label = "+- 1std")
plt.axvline(new_mean - new_std, linestyle = '--', color = "red")
plt.axvline(new_mean + new_std * 2, linestyle = '--', color = "purple", label = "+- 2std")
plt.axvline(new_mean - new_std * 2, linestyle = '--', color = "purple")
plt.axvline(new_mean + new_std * 3, linestyle = '--', color = "green", label = "+- 3 std")
plt.axvline(new_mean - new_std * 3, linestyle = '--', color = "green")
plt.title("Вероятность с выбросами распределения")
plt.ylabel("Вероятность")
plt.legend()
plt.show()

#Среднее значение немного увеличилось и стандартное отклонение также увеличилось. На втором графике справа видно,
#что в данных могут присутствовать выбросы из-за которых данные слегка сдвинуты от основного графика.
```

```python
np.random.seed(105)

CityA = np.random.normal(50000, 15000, 1000)
CityB = np.random.normal(55000, 20000, 1000)

mean_A = np.mean(CityA)
mean_B = np.mean(CityB)

sns.kdeplot(CityA, label = "CityA")
sns.kdeplot(CityB, label = "CityB")
plt.axvline(mean_A, linestyle = '--', color = "red", label = "Среднее CityA")
plt.axvline(mean_B, linestyle = '--', color = "green", label = "Среднее CityB")
plt.title("Распределение доходов жителей двух городов")
plt.xlabel("Доход")
plt.ylabel("Плотность")
plt.legend()
plt.show()
```

```python
np.random.seed(105)

population = np.random.exponential(10, 1000)

sample = np.random.choice(population, 100, False)

sns.kdeplot(sample)
plt.hist(sample, rwidth = 0.9, bins = 20, density = True)
plt.title("Распределение времени обслуживания")
plt.xlabel("Время обслуживания")
plt.ylabel("Плотность")
plt.show()

sm.qqplot(sample, line = 'r')
plt.show()
```
