```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
np.random.seed(105)
standart_normal_sample = np.random.normal(0, 1, 1000)
normal_sample = np.random.normal(8, 2, 1000)

mean = np.mean(normal_sample)
std = np.std(normal_sample)

plt.hist(standart_normal_sample, rwidth = 0.9, density = True, alpha = 0.5, bins = np.arange(-5, 10, 0.5), label = "Стандартное нормальное распределение")
plt.hist(normal_sample, rwidth = 0.9, density = True, alpha = 0.5, bins = np.arange(-5, 16, 0.5), label = "Нормальное распределение")
plt.legend()
plt.show()

normal_shifted_sample = (normal_sample - mean) / std

plt.hist(standart_normal_sample, rwidth = 0.9, density = True, alpha = 0.5, bins = np.arange(-5, 5, 0.5), label = "Стандартное нормальное распределение")
plt.hist(normal_shifted_sample, rwidth = 0.9, density = True, alpha = 0.5, bins = np.arange(-5, 5, 0.5), label = "Нормальное распределение")
plt.legend()
plt.show()
```

```python
np.random.seed(105)
height = np.random.normal(170, 10, 1000)
mean = np.mean(height)
std = np.std(height)

plt.hist(height, rwidth = 0.9, bins = 20)
plt.title("Распределение роста учащихся")
plt.show()

height_standart = (height - mean) / std

plt.hist(height_standart, rwidth = 0.9, bins = 20)
plt.title("Распределение стандартизированного роста")
plt.show()
```

```python
np.random.seed(105)
temperature = np.random.normal(22, 3, 1000)

high_temperature = temperature[temperature > 27]
low_temperature = temperature[temperature < 15]

print("Процент аномально высоких температур: ", len(high_temperature) / len(temperature) * 100)
print("Процент аномально низких температур: ", len(low_temperature) / len(temperature) * 100)

normal_temperature = temperature[(temperature >= 15) & (temperature <= 27)]

bins = np.arange(int(temperature.min()), int(temperature.max()) + 2, 1)

plt.hist(normal_temperature, rwidth = 0.9, bins = bins, color = "green", label = "Нормальная температура")
plt.hist(high_temperature, rwidth = 0.9, bins = bins, color = "red", label = "Аномально высокая температура")
plt.hist(low_temperature, rwidth = 0.9, bins = bins, color = "blue", label = "Аномально низкая температура")
plt.title("Среднесуточная температура для июля")
plt.legend()
plt.show()
```
