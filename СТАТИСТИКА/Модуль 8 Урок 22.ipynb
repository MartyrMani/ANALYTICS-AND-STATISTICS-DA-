## 1. Создайте колонку BMI (индекс массы тела), которая считается по формуле: BMI = (Вес в кг) / (Рост в метрах)^2. Учтите, что ваш текущий рост в см, не метры.

```python
data['BMI'] = data['Weight'] / (data['Height'] / 100) ** 2
data.head()
```

## 8. Есть ли в данных какие-то выбросы для роста, веса, BMI? Как можно их найти? Используя межквартильный размах? Удалите все такие выбросы из таблицы.

```python
q1_height = np.percentile(data['Height'], 25)
q3_height = np.percentile(data['Height'], 75)

iqr_height = q3_height - q1_height

lower_height = q1_height - 1.5 * iqr_height
upper_height = q3_height + 1.5 * iqr_height

print("Q1:", q1_height)
print("Q3:", q3_height)
print("IQR:", iqr_height)
print("Нижняя граница:", lower_height)
print("Верхняя граница:", upper_height)
```

```python
q1_weight = np.percentile(data['Weight'], 25)
q3_weight = np.percentile(data['Weight'], 75)

iqr_weight = q3_weight - q1_weight

lower_weight = q1_weight - 1.5 * iqr_weight
upper_weight = q3_weight + 1.5 * iqr_weight

print("Q1:", q1_weight)
print("Q3:", q3_weight)
print("IQR:", iqr_weight)
print("Нижняя граница:", lower_weight)
print("Верхняя граница:", upper_weight)
```

```python
q1_bmi = np.percentile(data['BMI'], 25)
q3_bmi = np.percentile(data['BMI'], 75)

iqr_bmi = q3_bmi - q1_bmi

lower_bmi = q1_bmi - 1.5 * iqr_bmi
upper_bmi = q3_bmi + 1.5 * iqr_bmi

print("Q1:", q1_bmi)
print("Q3:", q3_bmi)
print("IQR:", iqr_bmi)
print("Нижняя граница:", lower_bmi)
print("Верхняя граница:", upper_bmi)
```

```python
data = data[(data['Height'] >= lower_height) & (data['Height'] <= upper_height) &
    (data['Weight'] >= lower_weight) & (data['Weight'] <= upper_weight) &
    (data['BMI'] >= lower_bmi) & (data['BMI'] <= upper_bmi)]

data.shape
```
