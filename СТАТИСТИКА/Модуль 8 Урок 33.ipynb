```python
from scipy.stats import norm
import numpy as np
import random
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

```python
mean = 30
std = 1

threshold = 29

probability = norm.cdf(threshold, loc = mean, scale = std)

print(f"Вероятность что эспрессо менее {threshold} мл составляет: {probability:.4f}")
print("Есть ли здесь уровень значимости?: ", probability > 0.05)

#Можно увидеть подтверждение что эспрессо меньше 29 мл кофе взаправду происходит и это не является случайностью
```

```python
#Задача 3
np.random.seed(105)

random_data = np.random.normal(80, 20, 500).tolist() + (np.random.random(1000) * 100).tolist() + np.random.exponential(5, 200).tolist()

print("Среднее значение: ", np.mean(random_data))
```

```python
collected_means = []

for i in range(1000):
    sample_data = np.random.choice(random_data, 300)
    sample_mean = np.mean(sample_data)
    collected_means.append(sample_mean)

plt.hist(collected_means, rwidth = 0.9, density = True)
plt.show()

sm.qqplot(np.array(collected_means), line = "r")
plt.show()
```
