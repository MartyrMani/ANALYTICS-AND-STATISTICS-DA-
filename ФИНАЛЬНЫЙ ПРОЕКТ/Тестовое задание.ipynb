```python
ab_df = pd.read_excel('Данные для тестового задания.xlsx', sheet_name='Данные АБ тестов')

print(ab_df.head())

for exp in [1, 2, 3]:
    experiment = ab_df[ab_df['experiment_num'] == exp]

    control = experiment[experiment['experiment_group'] == 'control']['revenue']

    test = experiment[experiment['experiment_group'] == 'test']['revenue']

    print('Эксперимент', exp)
    print('ARPU control:', control.mean())
    print('ARPU test:', test.mean())

    t_stat, p_value = ttest_ind(
        control,
        test,
        equal_var=False
    )

    print('p-value:', p_value)
    print()
```
